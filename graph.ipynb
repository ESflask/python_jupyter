```python
import pandas as pd
import matplotlib.pyplot as plt
import matplotlib
matplotlib.rcParams['font.family'] = 'Meiryo'

df = pd.DataFrame({     # ここがグラフに入れるデータ本体
    '月': [1,2,3,4,5,6],
    '売上' : [120,150,130,170,160,200]
})
df.plot(
    x='月',
    y='売上',
    kind='line',        # グラフの種類(ここでは折れ線グラフ)
    title='月別売上推移',
    xlabel='月',        # x軸のラベル名
    ylabel='売上'       # y軸のラベル名
)
plt.show()  # グラフ表示
```

```python
df = pd.DataFrame({
    '月': [1,2,3,4,5,6],
    'りんご': [80,95,70,110,100,130],
    'みかん' : [60,55,80,75,90,100],
    'ブドウ' : [40,60,50,65,70,85]
})
a = ['りんご','みかん','ブドウ']    # 必ず[]で囲む
df.plot(
    x='月',
    y=a,
    kind='line',
    title='月別果物売上推移',
    xlabel='月',
    ylabel='売上'
)
plt.show()
```

```python
import pandas as pd
import matplotlib.pylot as plt

df = pd.DataFrame({
    '店舗' : ['東京','大阪','名古屋','福岡'],
    '売上' : [320,280,210,190]
})

df.plot(
    x = '店舗',
    y = '売上',
    kind = 'bar',
    title = '都市別売上',
    ylabel = '売上',
    color = 'steelblue',     # グラフの色
    edgecolor = 'black',
    width = 0.5,
    legend=False,
    rot=0
)
```

```python
import pandas as pd
import matplotlib.pyplot as plt

df =  pd.DataFrame({
    '部署' : ['営業','開発','総務','マーケ','サポート'],
    '残業時間' : [45,60,20,35,28]
})
df.plot(
    x = '部署',
    y = '残業時間',
    kind = 'barh',
    title = '部署別残業時間',
    xlabel = '時間(h)',
    ylabel = '部署',
    color = 'steelblue',
    width = 0.5,
    legend = False,
    rot = 0
)
plt.tight_layout()
```

```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.DataFrame({
    '月' : ['1月','2月','3月','4月'],
    'りんご' : [100,130,90,150],
    'みかん' : [80,60,110,70]
})

fig,axes = plt.subplots(
    nrows=1,
    ncols=2,
    figsize=(10,4)
)

df.plot(
    x='月',
    y='りんご',
    kind='bar',
    title='りんごの売上',
    ax=axes[0],
    color='tomato',
    legend=False
)
plt.tight_layout()
plt.show()
plt.rcParams['font.family'] = 'IPAexGothic'
```
