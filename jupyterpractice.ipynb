```python
total = 0
number = 1
while number <= 10: #numberが 9 以下のとき、下記の作業を繰り返す。
    total = total + number
    number = number + 1 #numberが10以上になるまでさっきのnumberに1ずつ足していくのを繰り返す。なのでnumberが2+1=3,3+1=4と増えていく。最後は11になるが一応10である
#number の箱の中身は、ループが1周するたびに、次々と新しい数字に上書きされていく。1周目はnumberが1のときに 0 + 1,2周目は1 + 2と、number自体に複数の数字は保存されないが、totalに合計は保存されている。totalは計算結果を保存、累積するための箱。
print(f"合計は:{total}")
```
